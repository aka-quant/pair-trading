Относительные значения

```python
//@version=5
indicator("Delta % and MAs (Pair Trading)", overlay=false, max_lines_count=500)

// === Inputs ===
baseSymbol = input.symbol("BINANCE:BTCUSDT", "Base Symbol")
quoteSymbol = input.symbol("BINANCE:ETHUSDT", "Quote Symbol")

lenLong  = input.int(100, "Long MA")
lenShort = input.int(10,  "Short MA")

// === Request Security ===
baseClose  = request.security(baseSymbol, timeframe.period, close)
quoteClose = request.security(quoteSymbol, timeframe.period, close)

// === Delta Calculation ===
delta     = baseClose / quoteClose
smaRef    = ta.sma(delta, lenLong)
delta_pct = (delta - smaRef) / smaRef * 100
smaLong   = ta.sma(delta_pct, lenLong)
smaShort  = ta.sma(delta_pct, lenShort)

// === Median and StdDev ===
medianVal = ta.median(delta_pct, 1000)
stdDevVal = ta.stdev(delta_pct, 1000)

// === Bands ===
// ±1σ
upperBand  = medianVal + stdDevVal
lowerBand  = medianVal - stdDevVal
// ±2σ
upperBand2 = medianVal + 2.0 * stdDevVal
lowerBand2 = medianVal - 2.0 * stdDevVal

// === Plot Panel ===
plot(delta_pct, title="Delta %", color=color.orange, linewidth=2)
plot(smaLong,   title="Long MA %",  color=color.blue,   linewidth=2)
plot(smaShort,  title="Short MA %", color=color.red,    linewidth=2)

// ±1σ
plot(upperBand, title="Median + 1σ", color=color.green,  linewidth=2)
plot(lowerBand, title="Median - 1σ", color=color.maroon, linewidth=2)

// ±2σ (обычные линии)
plot(upperBand2, title="Median + 2σ", color=color.new(color.green, 0),  linewidth=2)
plot(lowerBand2, title="Median - 2σ", color=color.new(color.maroon, 0), linewidth=2)
```
